```python
a=[10,20,30]
print(type(a))
lst = [10, 20, 30, 40]
print("Initial list:", lst)
print("First element:", lst[0])
print("Last element:", lst[-1])
lst.append(50)
print("After append:", lst)
lst.insert(2, 25)
print("After insert:", lst)
lst.extend([60, 70])
print("After extend:", lst)
lst.remove(25)
print("After remove:", lst)
lst.pop()
print("After pop:", lst)
lst.pop(1)
print("After pop at index 1:", lst)
lst[0] = 100
print("After updating:", lst)
print("Index of 30:", lst.index(30))
print("Count of 20:", lst.count(20))

```

```python
# DICTIONARY OPERATIONS
d = {
    "name": "G.Thanmayee",
    "age": 19,
    "course": "BTech"
}
print("Original dictionary:", d)
print("Name:", d["name"])
print("Age:", d.get("age"))
d["college"] = "XYZ University"
print("After adding:", d)
d["age"] = 21
print("After updating:", d)
d.pop("course")
print("After pop:", d)
removed_item = d.popitem()
print("After popitem:", d)
print("Removed item:", removed_item)
print("Keys:", d.keys())
print("Values:", d.values())
print("Items:", d.items())
print("Is 'name' a key?", "name" in d)
print("Length of dictionary:", len(d))
# Nested dictionary
student = {
    "id": 1,
    "info": {
        "name": "Amit",
        "marks": 70
    }
}
print("Nested dictionary value:", student["info"]["marks"])

# Clearing dictionary
d.clear()
print("After clear:", d)
```

```python
# SET OPERATIONS
s1 = {10, 20, 30}
s2 = {30, 40, 50}
print("Union using method:", s1.union(s2))
print("Intersection using method:", s1.intersection(s2))
print("Difference using method:", s1.difference(s2))
print("Symmetric difference using method:", s1.symmetric_difference(s2))
```

```python
# TUPLE OPERATIONS
t = (10, 20, 30, 40)
print("Original tuple:", t)
print("First element:", t[0])
print("Last element:", t[-1])
print("Sliced tuple:", t[1:4])
print("Length of tuple:", len(t))
print("Count of 20:", t.count(20))
print("Index of 40:", t.index(40))
t2 = (50, 60)
t3 = t + t2
print("After concatenation:", t3)
```

```python
# Create and write to a file
file_path = "sample_text.txt"

with open(file_path, "w") as file:
    file.write("Hello World\n")
    file.write("This is a Python file handling example.\n")
    file.write("We are counting lines, words, and characters.\n")
print("File created and data written successfully.\n")
# Read the file
with open(file_path, "r") as file:
    content = file.read()
print("File Content:\n")
print(content)
lines = content.split("\n")
num_lines = len([line for line in lines if line.strip() != ""])
num_words = len(content.split())
num_characters = len(content)
print("Number of lines:", num_lines)
print("Number of words:", num_words)
print("Number of characters:", num_characters)
```
