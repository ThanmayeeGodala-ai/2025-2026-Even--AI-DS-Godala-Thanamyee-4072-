```python
#35 networkx library & SNA
import networkx as nx
import matplotlib.pyplot as plt
G=nx.Graph()
G.add_edges_from([('A','B'),('A','C'),('B','C'),('C','D')])
plt.figure(figsize=(2,2))
nx.draw(G,with_labels=True)
plt.title("Social Network Analysis")
plt.show()
print("Degree of centrality:",nx.degree_centrality(G))
print("Closeness centrality:",nx.closeness_centrality(G))
print("Betweenness centrality:",nx.betweenness_centrality(G))
```

```python
#36
import networkx as nx
import matplotlib.pyplot as plt
G=nx.Graph()
G.add_edges_from([('A','B'),('A','C'),('B','D'),('C','D'),('E','F'),('F','H'),('G','H'),('D','E')])
from networkx.algorithms.community import greedy_modularity_communities
communities=greedy_modularity_communities(G)
for i,com in enumerate(communities):
    print(f"community {i+1}:{list(com)}")
nx.draw(G,with_labels=True,node_size=3000)
plt.figure(figsize=(2,1))
plt.show()
```

```python
#37
import networkx as nx
G=nx.DiGraph()
G.add_edges_from([('A','B'),('A','C'),('B','C'),('C','D'),('E','B'),('E','C')])
pagerank=nx.pagerank(G,alpha=0.85)
print("pagerank:")
for page,rank in pagerank.items():
    print(f"{page}:{rank:.4f}")
```
