```python
#descriptive statistic function
import pandas as pd
import numpy as np
data={
    "age":[22,25,47,35,46,50,28,33,39,41],
    "salary":[3000,3200,5500,4000,5200,6000,3500,3700,4500,4800]
}
df=pd.DataFrame(data)
print("Summary Statistics:")
print("mean:",df.mean())
print("median:",df.median())
print("mode:",df.mode())
print("\n minimum value:",df.min())
print("\n maximum value:",df.max())
print("\n standard deviation:",df.std())
print("\n variance:",df.var())
print("\n count:",df.count())
print("\n skewness:",df.skew())
print("\n kurtosis:",df.kurtosis())
```

```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, None, 35, 40, 28],
    'Salary': [50000, 60000, None, 70000, 62000],
    'Allowances': [2000, None, 2000, 2000, 2500]
}
df = pd.DataFrame(data)
print("Missing Values Count:\n", df.isnull().sum())
df_cleaned = df.dropna()
print("\nDataFrame after dropping missing values:\n", df_cleaned)
df_filled = df.fillna({'Age': df['Age'].mean(), 'Salary': df['Salary'].mean(), 'Allowances': 2000})
print("\nDataFrame after filling missing values:\n", df_filled)
```

```python
#IQR meethod
import pandas as pd
data={
    "name":['A','B','C','D','E','F'],
    "age":[13,24,25,120,23,22]

}

df=pd.DataFrame(data)
q1=df['age'].quantile(0.25)
q3=df['age'].quantile(0.75)
iqr=q3-q1
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr
outliners=df[(df['age']<lower_bound) | (df['age']>upper_bound)]
print("outliners using IQR method")
print(outliners)
```

```python
# Data Distribution

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
normal_data = np.random.normal(loc=70, scale=5, size=1000)
right_skewed = np.random.exponential(scale=2, size=1000)
left_skewed = -np.random.exponential(scale=2, size=1000)
plt.hist(normal_data, bins=30)
plt.title("Normal Distribution")
plt.show()
plt.hist(right_skewed, bins=30)
plt.title("Right Skewed Distribution")
plt.show()
plt.hist(left_skewed, bins=30)
plt.title("Left Skewed Distribution")
plt.show()
print("Normal Data Mean:", np.mean(normal_data))
print("Normal Data Median:", np.median(normal_data))

print("Right Skewed Mean:", np.mean(right_skewed))
print("Right Skewed Median:", np.median(right_skewed))

print("Left Skewed Mean:", np.mean(left_skewed))
print("Left Skewed Median:", np.median(left_skewed))
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
print(" Original Dataset Info ")
print(df.info())
print("\n Missing Values Before Cleaning:")
print(df.isnull().sum())
print("\nFirst 5 rows:\n", df.head())
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
print("\nMissing Values After Filling:")
print(df.isnull().sum())
df = df.drop(columns=['Cabin'])
print("\n Columns After Dropping 'Cabin' ")
print(df.columns)
before_duplicates = df.shape[0]
df = df.drop_duplicates()
after_duplicates = df.shape[0]
print(f"\nRemoved {before_duplicates - after_duplicates} duplicate rows")
df['Survived'] = df['Survived'].astype('category')
df['Pclass'] = df['Pclass'].astype('category')
print("\n Data Types After Conversion ")
print(df.dtypes)
df.columns = df.columns.str.lower()
print("\n Columns After Renaming ")
print(df.columns)
Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
before_outliers = df.shape[0]
df = df[(df['age'] >= lower_bound) & (df['age'] <= upper_bound)]
after_outliers = df.shape[0]
print(f"\nRemoved {before_outliers - after_outliers} outlier rows based on Age")
df.to_csv("cleaned_titanic.csv", index=False)
print("\nData Cleaning Completed Successfully!")
print("Cleaned file saved as: cleaned_titanic.csv")
print("\n First 5 rows of Cleaned Dataset ")
print(df.head())
```
